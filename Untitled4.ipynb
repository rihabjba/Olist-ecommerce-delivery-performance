```python
import sqlite3
import pandas as pd

# 1. Read the CSV files you just uploaded
customers = pd.read_csv('olist_customers_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')

# 2. Create a temporary local database
conn = sqlite3.connect('ecommerce.db')

# 3. Put our files into SQL tables so we can query them
customers.to_sql('customers', conn, index=False, if_exists='replace')
orders.to_sql('orders', conn, index=False, if_exists='replace')

# 4. Let's run a test query to see if it worked!
test_query = "SELECT * FROM customers LIMIT 5;"
test_result = pd.read_sql(test_query, conn)

# 5. Show the result
display(test_result)
```

```python
query_1 = """
    SELECT
        c.customer_state,
        COUNT(o.order_id) AS total_orders
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    GROUP BY c.customer_state
    ORDER BY total_orders DESC
    LIMIT 10;
"""

top_states = pd.read_sql(query_1, conn)
display(top_states)
```

```python
top_states.to_csv('top_states_orders.csv', index=False)
print("File saved successfully!")
```

# **The Late Delivery Query**




```python
query_2 = """
    SELECT
        CASE
            WHEN order_delivered_customer_date > order_estimated_delivery_date THEN 'Late'
            ELSE 'On Time'
        END AS delivery_status,
        COUNT(order_id) AS total_orders
    FROM orders
    WHERE order_status = 'delivered'
      AND order_delivered_customer_date IS NOT NULL
    GROUP BY delivery_status;
"""

delivery_performance = pd.read_sql(query_2, conn)
display(delivery_performance)

# Save it for our dashboard later!
delivery_performance.to_csv('delivery_performance.csv', index=False)
```

```python
query_3 = """
    SELECT
        c.customer_state,
        COUNT(o.order_id) AS total_orders,
        SUM(CASE WHEN o.order_delivered_customer_date > o.order_estimated_delivery_date THEN 1 ELSE 0 END) AS late_orders
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    WHERE o.order_status = 'delivered'
      AND o.order_delivered_customer_date IS NOT NULL
    GROUP BY c.customer_state
    ORDER BY late_orders DESC
    LIMIT 10;
"""

state_delays = pd.read_sql(query_3, conn)
display(state_delays)

# Save this for our dashboard too!
state_delays.to_csv('state_delays.csv', index=False)
```
